#IMPORTING MODULES 👎🏻
In Python, modules are files that contain reusable code (functions, classes, or variables) which you can import into your program to avoid rewriting logic. They help organize projects, make code cleaner, and allow you to use built-in or third‑party libraries like math, os, numpy, or your own custom .py files.

```python
# 1. Importing an entire module
import math

# Using functions from the imported math module
print("Square root of 16:", math.sqrt(16))
print("Value of Pi:", math.pi)

# 2. Importing specific functions from a module
from random import randint, choice

# Generates a random integer between 1 and 10
random_num = randint(1, 10)
print("\nRandom number between 1 and 10:", random_num)

# Picks a random item from a list
items = ['rock', 'paper', 'scissors']
print("Random choice:", choice(items))
```

variable is like a container that stores data (numbers, text, lists, etc.).
#TYPE OF VARIABLE
Local Variable → Defined inside a function, usable only there.

Global Variable → Defined outside functions, usable everywhere.

```python
# 1. Global Variable (accessible anywhere in the script)
global_var = "I am a global variable"

def my_function():
    # 2. Local Variable (accessible only inside this function)
    local_var = "I am a local variable"
    print("Inside function:", global_var)  # Accessing global variable
    print("Inside function:", local_var)   # Accessing local variable

my_function()

# Accessing variables outside the function
print("\nOutside function:", global_var)

try:
    print(local_var)  # This will raise an error because local_var is restricted to the function
except NameError as e:
    print("Error outside function:", e)
```

#LIST IN PYTHON 👎🏻

```python
# Creating a list
my_list = [10, 20, "apple", 3.14, True]
print("Original list:", my_list)

# Accessing elements (0-indexed)
print("First element:", my_list[0])
print("Last element:", my_list[-1])

# Modifying an element
my_list[1] = 50
print("Modified list:", my_list)

# Appending a new element
my_list.append("banana")
print("After append:", my_list)

# Removing an element
my_list.remove("apple")
print("After removing 'apple':", my_list)

# Slicing a list
print("Sliced list (first 3 elements):", my_list[:3])
```

#TUPLE IN PYTHON 👎🏻

```python
# Creating a tuple
my_tuple = (10, 20, "apple", 3.14, True)
print("Original tuple:", my_tuple)

# Accessing elements (0-indexed)
print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])

# Tuples are immutable (cannot be modified)
try:
    my_tuple[1] = 50
except TypeError as e:
    print("Error when trying to modify element:", e)

# Slicing a tuple
print("Sliced tuple (first 3 elements):", my_tuple[:3])

# Finding length
print("Length of tuple:", len(my_tuple))
```

DICTIONARY IN PYTHON 👎🏻

```python
# Creating a dictionary
my_dict = {
    "name": "Yashoda",
    "age": 25,
    "city": "New York",
    "skills": ["Python", "Data Science"]
}
print("Original dictionary:", my_dict)

# Accessing values using keys
print("Name:", my_dict["name"])
print("Age (using get()):", my_dict.get("age"))

# Modifying/Updating a value
my_dict["age"] = 26
print("Updated Age:", my_dict["age"])

# Adding a new key-value pair
my_dict["profession"] = "Developer"
print("After adding profession:", my_dict)

# Removing a key-value pair
removed_value = my_dict.pop("city")
print("Removed city:", removed_value)
print("After popping city:", my_dict)

# Getting keys and values
print("Keys:", list(my_dict.keys()))
print("Values:", list(my_dict.values()))
print("Items:", list(my_dict.items()))
```

#SETS IN PYTHON 👎🏻

```python
# Creating a set (unordered collection of unique items)
my_set = {1, 2, 3, 3, 4, 5}
print("Original Set (duplicates removed automatically):", my_set)

# Adding elements
my_set.add(6)
print("After adding 6:", my_set)

# Removing elements
my_set.discard(2)
print("After discarding 2:", my_set)

# Set Operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union (all unique elements from both sets)
print("Union:", set_a.union(set_b))

# Intersection (elements present in both sets)
print("Intersection:", set_a.intersection(set_b))

# Difference (elements in set_a but not in set_b)
print("Difference (A - B):", set_a.difference(set_b))
```

### **For Loops in Python**

A `for` loop is used to iterate over a sequence (such as a list, tuple, dictionary, set, or string) or a range of numbers. It executes a block of code repeatedly for each item in the sequence.

```python
# 1. Looping through a range of numbers using range()
print("Looping from 0 to 4:")
for i in range(5):
    print(i)

# 2. Looping through a list
fruits = ["apple", "banana", "cherry"]
print("\nLooping through a list:")
for fruit in fruits:
    print(f"I like {fruit}")

# 3. Looping through a string
print("\nLooping through a string:")
for letter in "Python":
    print(letter)

# 4. Looping through a dictionary (keys and values)
person = {"name": "Yashoda", "role": "Developer"}
print("\nLooping through a dictionary:")
for key, value in person.items():
    print(f"{key}: {value}")
```

### **While Loops in Python**

A `while` loop repeatedly executes a block of code as long as a specified condition remains `True`.

* **Crucial Rule:** You must update the loop variable inside the block so that the condition eventually becomes `False`. Otherwise, you will create an **infinite loop**!

```python
# 1. Basic while loop
print("Basic while loop count:")
count = 1
while count <= 5:
    print(count)
    count += 1  # Increment to eventually make the condition False

# 2. Using 'break' to exit a loop early
print("\nUsing break statement:")
i = 1
while i < 10:
    print(i)
    if i == 3:
        print("Breaking out of the loop!")
        break
    i += 1

# 3. Using 'continue' to skip the current iteration
print("\nUsing continue statement to skip even numbers:")
n = 0
while n < 6:
    n += 1
    if n % 2 == 0:
        continue  # Skip the rest of the code block for even numbers
    print(n)
```

#FUNCTIONS IN PYTHON 👎🏻

```python
# 1. Defining a basic function
def greet_user():
    print("Hello! Welcome to the world of Python functions.")

# Calling the function
greet_user()

# 2. Function with parameters and a return value
def add_numbers(num1, num2):
    return num1 + num2

result = add_numbers(15, 25)
print("Result of addition:", result)

# 3. Function with a default parameter value
def greet_person(name="Guest"):
    print(f"Hello, {name}!")

greet_person("Yashoda")
greet_person()  # Uses the default value 'Guest'
```

EXPLANATION OF TRY AND EXECPT BLOCKS

```python
# Exception handling in Python prevents the program from crashing when an error occurs.

# 1. Basic Try-Except Block
try:
    # Code that might cause an error
    result = 10 / 0
except ZeroDivisionError:
    # Code that runs if the specific error occurs
    print("Error: You cannot divide a number by zero!")

# 2. Handling multiple types of exceptions
try:
    number = int("not_a_number")  # This raises a ValueError
except ValueError as e:
    print(f"Value Error caught: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")

# 3. Using Else and Finally blocks
try:
    print("\nTrying to divide 10 by 2...")
    val = 10 / 2
except ZeroDivisionError:
    print("Can't divide by zero!")
else:
    # Runs ONLY if the try block succeeds without any exceptions
    print(f"Division was successful! Result: {val}")
finally:
    # This block ALWAYS runs, regardless of whether an exception occurred or not
    print("This 'finally' block always executes for cleanup purposes.")
```
